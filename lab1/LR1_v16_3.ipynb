**Задание 1.** Создать класс String (строка). Разработать в классе методы для работы со строками (не менее пяти).


```python
def upper_char(ch):
    code = ord(ch)
    if ord("a") <= code <= ord("z"):
        return chr(code - (ord("a") - ord("A")))
    if ord("а") <= code <= ord("я"):
        return chr(code - (ord("а") - ord("А")))
    if ch == "ё":
        return "Ё"
    return ch


def lower_char(ch):
    code = ord(ch)
    if ord("A") <= code <= ord("Z"):
        return chr(code + (ord("a") - ord("A")))
    if ord("А") <= code <= ord("Я"):
        return chr(code + (ord("а") - ord("А")))
    if ch == "Ё":
        return "ё"
    return ch


class String:
    def __init__(self, text):
        self.text = text

    def length(self):
        count = 0
        for _ in self.text:
            count += 1
        return count

    def reverse(self):
        length = self.length()
        result = ""
        i = length - 1
        while i >= 0:
            result += self.text[i]
            i -= 1
        return result

    def to_upper(self):
        result = ""
        for ch in self.text:
            result += upper_char(ch)
        return result

    def to_lower(self):
        result = ""
        for ch in self.text:
            result += lower_char(ch)
        return result

    def is_palindrome(self):
        cleaned = ""
        for ch in self.text:
            if ch != " ":
                cleaned += lower_char(ch)
        length = 0
        for _ in cleaned:
            length += 1
        reversed_text = ""
        i = length - 1
        while i >= 0:
            reversed_text += cleaned[i]
            i -= 1
        return cleaned == reversed_text

    def count_char(self, char):
        count = 0
        for ch in self.text:
            if ch == char:
                count += 1
        return count

    def replace_char(self, old, new):
        result = ""
        for ch in self.text:
            if ch == old:
                result += new
            else:
                result += ch
        return result

    def words(self):
        result = []
        current = ""
        for ch in self.text + " ":
            if ch != " ":
                current += ch
            else:
                if current != "":
                    result.append(current)
                    current = ""
        return result

    def without_punctuation(self):
        punctuation = ".,!?;:()[]{}\"'«»—–…-"
        result = ""
        for ch in self.text:
            if ch not in punctuation:
                result += ch
        return result


sample = String("А роза упала на лапу Азора")
print("Исходная строка:", sample.text)
print("1) Длина строки:", sample.length())
print("2) Строка в обратном порядке:", sample.reverse())
print("3) Верхний регистр:", sample.to_upper())
print("4) Нижний регистр:", sample.to_lower())
print("5) Это палиндром:", sample.is_palindrome())
print("6) Количество символов «а»:", sample.count_char("а"))
print("7) Замена пробелов на «_»:", sample.replace_char(" ", "_"))
print("8) Слова строки:", sample.words())

other = String("Привет, мир!")
print()
print("Вторая строка:", other.text)
print("9) Без знаков препинания:", other.without_punctuation())
print("10) Это палиндром:", other.is_palindrome())
```

**Задание 2.** Создать класс TrafficLight (светофор). Определить у него один атрибут color (цвет) и метод running (запуск); атрибут реализовать как приватный; в рамках метода реализовать переключение светофора в режимы красный, жёлтый, зелёный; продолжительность первого состояния (красный) — 7 секунд, второго (жёлтый) — 2 секунды, третьего (зелёный) — 4 секунды; переключение между режимами должно осуществляться только в указанном порядке; проверить работу примера, создав экземпляр и вызвав описанный метод. При нарушении порядка режимов выводить соответствующее сообщение и завершать скрипт.


```python
import sys
import time


class TrafficLight:
    __states = ("красный", "жёлтый", "зелёный")
    __durations = {"красный": 7, "жёлтый": 2, "зелёный": 4}

    def __init__(self, color="красный"):
        if color not in TrafficLight.__states:
            raise ValueError("неизвестный режим: " + color)
        self.__color = color

    def current_color(self):
        return self.__color

    def __next_color(self):
        index = 0
        i = 0
        while i < len(TrafficLight.__states):
            if TrafficLight.__states[i] == self.__color:
                index = i
            i += 1
        return TrafficLight.__states[(index + 1) % len(TrafficLight.__states)]

    def switch(self, new_color):
        if new_color not in TrafficLight.__states:
            print("Ошибка: неизвестный режим", new_color)
            print("Работа светофора остановлена.")
            sys.exit(1)
        expected = self.__next_color()
        if new_color != expected:
            print("Ошибка: нарушен порядок переключения.")
            print("После режима «" + self.__color + "» должен следовать «" + expected + "», а получен «" + new_color + "».")
            print("Работа светофора остановлена.")
            sys.exit(1)
        self.__color = new_color
        duration = TrafficLight.__durations[new_color]
        print("Включён режим:", new_color, "— продолжительность", duration, "с")
        time.sleep(duration)

    def running(self, cycles):
        print("Светофор включён. Начальный режим:", self.__color)
        number = 0
        while number < cycles:
            self.switch("жёлтый")
            self.switch("зелёный")
            self.switch("красный")
            number += 1
            print("Выполнено полных циклов:", number)


try:
    light = TrafficLight()
    light.running(2)
    print()
    print("Проверка контроля порядка переключения:")
    print("Текущий режим:", light.current_color())
    light.switch("зелёный")
    print("Эта строка не выполняется: скрипт завершён при нарушении порядка")
except ValueError as error:
    print("Ошибка:", error)
finally:
    print("finally: работа со светофором завершена")
```

**Задание 3.** Создать классы «Зоомагазин», «Животное», «Рыбы», «Птицы». Определить свойства породу и стоимость для указанных животных (рыб, птиц), в каждом классе реализовать переопределение метода «способ передвижения». Вывести данные о самой дорогой породе. Предусмотреть метод записи информации в файл.


```python
import os


class Animal:
    def __init__(self, breed, price):
        self.breed = breed
        self.price = price

    def kind(self):
        return "животное"

    def move(self):
        return "передвигается по земле"

    def info(self):
        return self.kind() + " породы «" + self.breed + "», стоимость " + str(self.price) + " руб."


class Fish(Animal):
    def kind(self):
        return "рыба"

    def move(self):
        return "плавает в воде"


class Bird(Animal):
    def kind(self):
        return "птица"

    def move(self):
        return "летает по воздуху"


class PetShop:
    def __init__(self, title):
        self.title = title
        self.animals = []

    def add(self, animal):
        self.animals.append(animal)

    def show(self):
        for animal in self.animals:
            print(animal.info() + " — способ передвижения: " + animal.move())

    def most_expensive(self):
        best = None
        for animal in self.animals:
            if best is None or animal.price > best.price:
                best = animal
        return best

    def save_to_file(self, file_name):
        with open(file_name, "w", encoding="utf-8") as file:
            file.write("Зоомагазин «" + self.title + "»\n")
            for animal in self.animals:
                file.write(animal.info() + "; способ передвижения: " + animal.move() + "\n")
            best = self.most_expensive()
            file.write("Самая дорогая порода: " + best.breed + " — " + str(best.price) + " руб.\n")
        return file_name


FILE_NAME = os.path.join(os.path.dirname(os.path.abspath(__file__)), "zoo_shop.txt")

shop = PetShop("Аквариум и птицы")
shop.add(Fish("гуппи", 12.50))
shop.add(Fish("скалярия", 24.90))
shop.add(Bird("волнистый попугай", 65.00))
shop.add(Bird("канарейка", 58.40))
shop.add(Bird("попугай ара", 145.00))
shop.add(Animal("морская свинка", 35.00))

print("Зоомагазин:", shop.title)
print("Животные и способы передвижения:")
shop.show()

best = shop.most_expensive()
print()
print("Самая дорогая порода:", best.breed)
print("Данные о ней:", best.info(), "— способ передвижения:", best.move())

try:
    file_name = shop.save_to_file(FILE_NAME)
    print()
    print("Информация записана в файл:", os.path.basename(file_name))
    print("Содержимое файла:")
    with open(file_name, "r", encoding="utf-8") as file:
        for line in file:
            print(line.rstrip())
except OSError as error:
    print("Ошибка записи в файл:", error)
finally:
    print("finally: задание 3 части 3 завершено")
```

**Задание 4.** Придумать класс самостоятельно, реализовать в нём методы экземпляра класса, статические методы, методы класса.


```python
class Toy:
    shop_name = "Магазин игрушек"
    created = 0

    def __init__(self, title, price, count):
        self.title = title
        self.price = price
        self.count = count
        Toy.created += 1

    def cost(self, pieces):
        return self.price * pieces

    def info(self):
        print(self.title, ": цена", self.price, "руб., в наличии", self.count, "шт.")

    def buy(self, pieces):
        if pieces <= 0:
            raise ValueError("количество должно быть положительным")
        if pieces > self.count:
            raise ValueError("недостаточно товара, остаток " + str(self.count) + " шт.")
        self.count -= pieces
        return self.cost(pieces)

    @staticmethod
    def is_valid_price(value):
        return value > 0

    @staticmethod
    def boxes_needed(pieces, per_box):
        boxes = pieces // per_box
        if pieces % per_box != 0:
            boxes += 1
        return boxes

    @classmethod
    def from_dict(cls, data):
        return cls(data["title"], data["price"], data["count"])

    @classmethod
    def created_count(cls):
        return cls.created


doll = Toy("кукла", 45.5, 12)
car = Toy.from_dict({"title": "машинка", "price": 32.0, "count": 20})
doll.info()
car.info()
print("Стоимость 3 кукол:", doll.cost(3))
print("Цена 100 корректна:", Toy.is_valid_price(100))
print("Цена -5 корректна:", Toy.is_valid_price(-5))
print("Коробок для 20 машинок по 6 шт.:", Toy.boxes_needed(20, 6))

try:
    print("Покупка 3 кукол, стоимость:", doll.buy(3))
    print("Покупка 50 машинок:")
    print(car.buy(50))
except ValueError as error:
    print("Ошибка:", error)
finally:
    print("finally: задание 4 части 3 завершено")

print("Создано объектов класса Toy:", Toy.created_count())
print("Название магазина:", Toy.shop_name)
```
