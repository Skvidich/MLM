**Задание 1.** Простейший калькулятор с введёнными двумя числами вещественного типа. Ввод с клавиатуры операций + - * / и двух чисел. Операции являются функциями. Обработать ошибку «Деление на ноль». Ноль использовать в качестве завершения программы (сделать как отдельную операцию).


```python
def add(a, b):
    return a + b


def sub(a, b):
    return a - b


def mul(a, b):
    return a * b


def div(a, b):
    if b == 0:
        raise ZeroDivisionError("делитель равен нулю")
    return a / b


def calculate(a, op, b):
    if op == "+":
        return add(a, b)
    if op == "-":
        return sub(a, b)
    if op == "*":
        return mul(a, b)
    if op == "/":
        return div(a, b)
    raise ValueError("неизвестная операция " + op)


def run_calculator():
    print("Операции: + - * / . Ноль вместо числа или ноль вместо операции завершает программу.")
    operations_done = 0
    while True:
        try:
            a = float(input("Первое число (0 — завершение): "))
            print("Введено:", a)
            if a == 0:
                print("Получен ноль — завершение программы")
                break
            op = input("Операция (+ - * / , 0 — завершение): ").strip()
            print("Введено:", op)
            if op == "0":
                print("Получен ноль — завершение программы")
                break
            b = float(input("Второе число: "))
            print("Введено:", b)
            result = calculate(a, op, b)
            operations_done += 1
            print("Результат:", result)
        except ZeroDivisionError as error:
            print("Ошибка: деление на ноль (" + str(error) + ")")
        except ValueError as error:
            print("Ошибка: не удалось разобрать число или операцию —", error)
        finally:
            print("finally: итерация калькулятора завершена, выполнено операций:", operations_done)
    print("Всего выполнено операций:", operations_done)


try:
    run_calculator()
except Exception as error:
    print("Ошибка:", error)
finally:
    print("finally: задание 1 части 2 завершено")
```

**Задание 2.** Напишите функцию, которая будет принимать один аргумент. Если в функцию передаётся список — найти сумму чётных чисел, удалить все нулевые элементы. Если множество — удалить максимальный элемент. Число — определить, простое или нет. Строка — вывести на экран наиболее часто встречающийся символ. Сделать проверку со всеми этими случаями.


```python
def sum_even_and_remove_zeros(values):
    even_sum = 0
    for value in values:
        if value % 2 == 0:
            even_sum += value
    cleaned = []
    for value in values:
        if value != 0:
            cleaned.append(value)
    return even_sum, cleaned


def remove_maximum(values):
    maximum = None
    for value in values:
        if maximum is None or value > maximum:
            maximum = value
    values.discard(maximum)
    return maximum


def is_prime(number):
    if number < 2:
        return False
    divisor = 2
    while divisor * divisor <= number:
        if number % divisor == 0:
            return False
        divisor += 1
    return True


def most_frequent_char(text):
    counts = {}
    for ch in text:
        if ch in counts:
            counts[ch] = counts[ch] + 1
        else:
            counts[ch] = 1
    best_char = ""
    best_count = 0
    for ch in text:
        if counts[ch] > best_count:
            best_count = counts[ch]
            best_char = ch
    return best_char, best_count


def process_value(value):
    try:
        if type(value) is list:
            even_sum, cleaned = sum_even_and_remove_zeros(value)
            print("Список. Сумма чётных чисел:", even_sum)
            print("Список без нулевых элементов:", cleaned)
        elif type(value) is set:
            removed = remove_maximum(value)
            print("Множество. Удалён максимальный элемент:", removed)
            print("Множество после удаления:", value)
        elif type(value) is int:
            if is_prime(value):
                print("Число", value, "— простое")
            else:
                print("Число", value, "— составное")
        elif type(value) is str:
            char, count = most_frequent_char(value)
            print("Строка. Наиболее часто встречающийся символ:", char, "— вхождений:", count)
        else:
            print("Неподдерживаемый тип:", type(value))
    except Exception as error:
        print("Ошибка:", error)
    finally:
        print("finally: обработка значения завершена\n")


list_example = [4, 0, 7, 10, -6, 0, 3]
print("Проверка списка:", list_example)
process_value(list_example)

set_example = {5, 12, 7, 3}
print("Проверка множества:", set_example)
process_value(set_example)

print("Проверка числа: 97")
process_value(97)

print("Проверка числа: 100")
process_value(100)

print("Проверка строки: молоко")
process_value("молоко")
```

**Задание 3.** Дана целочисленная квадратная матрица. Определить сумму элементов в тех столбцах, которые не содержат отрицательных элементов.


```python
def check_row(numbers, size):
    if len(numbers) != size:
        raise ValueError("в строке должно быть чисел: " + str(size) + ", получено: " + str(len(numbers)))
    return numbers


def read_row(row_number, size):
    raw = input("Строка " + str(row_number) + " (" + str(size) + " чисел через пробел): ")
    print("Введено:", raw)
    numbers = []
    current = ""
    for ch in raw + " ":
        if ch != " ":
            current += ch
        else:
            if current != "":
                numbers.append(int(current))
                current = ""
    return check_row(numbers, size)


def read_matrix(size):
    matrix = []
    for i in range(size):
        matrix.append(read_row(i + 1, size))
    return matrix


def print_matrix(matrix):
    for row in matrix:
        line = ""
        i = 0
        while i < len(row):
            if i > 0:
                line += " "
            line += str(row[i])
            i += 1
        print(line)


def column_sums(matrix):
    size = len(matrix)
    sums = []
    for column in range(size):
        has_negative = False
        column_sum = 0
        for row in range(size):
            value = matrix[row][column]
            if value < 0:
                has_negative = True
            column_sum += value
        if has_negative:
            sums.append(None)
        else:
            sums.append(column_sum)
    return sums


def run_calculation():
    size = int(input("Размер квадратной матрицы n: "))
    print("Введено:", size)
    if size <= 0:
        raise ValueError("размер матрицы должен быть положительным")
    matrix = read_matrix(size)
    print("Введённая матрица:")
    print_matrix(matrix)

    sums = column_sums(matrix)
    total = 0
    print("Результат по столбцам:")
    for column in range(len(sums)):
        if sums[column] is None:
            print("  столбец", column + 1, "— содержит отрицательный элемент, не учитывается")
        else:
            print("  столбец", column + 1, "— сумма элементов:", sums[column])
            total += sums[column]
    print("Сумма элементов в столбцах без отрицательных элементов:", total)

    print("\nПроверка обработки исключительной ситуации: строка из трёх чисел вместо", size)
    try:
        check_row([1, 2, 3], size)
    except ValueError as error:
        print("Ошибка:", error)


try:
    run_calculation()
except ValueError as error:
    print("Ошибка:", error)
finally:
    print("finally: задание 3 части 2 завершено")
```

**Задание 4.** Написать программу, демонстрирующую работу try/except/finally.


```python
def divide(a, b):
    result = None
    try:
        print("try: делим", a, "на", b)
        result = a / b
        print("Результат:", result)
    except ZeroDivisionError:
        print("except: деление на ноль")
    except TypeError:
        print("except: аргументы должны быть числами")
    finally:
        print("finally: выполняется всегда, даже если была ошибка")
        print("finally: result =", result)
    return result


print("--- Успешное деление ---")
divide(20, 4)
print()
print("--- Деление на ноль ---")
divide(20, 0)
print()
print("--- Неверный тип ---")
divide(20, "четыре")
print()

try:
    a = int(input("Первое число: "))
    print("Введено:", a)
    b = int(input("Второе число: "))
    print("Введено:", b)
    divide(a, b)
except ValueError:
    print("except: нужно ввести целые числа")
finally:
    print("finally: задание 4 части 2 завершено")
```
