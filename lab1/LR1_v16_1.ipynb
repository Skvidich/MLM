**Задание 1.** На обработку поступает натуральное число. Написать программу, которая выводит на экран сумму чётных цифр этого числа или 0, если чётных цифр в записи нет.


```python
number = int(input("Введите натуральное число: "))
print("Введено:", number)

n = number
even_sum = 0
even_count = 0
while n > 0:
    digit = n % 10
    if digit % 2 == 0:
        even_sum += digit
        even_count += 1
    n //= 10

if even_count == 0:
    print("Чётных цифр в записи нет, результат равен 0")
else:
    print("Найдено чётных цифр:", even_count)
    print("Сумма чётных цифр:", even_sum)
```

**Задание 2.** Ввести строку текста. Вывести на экран самое длинное слово. Все буквы нижнего регистра перевести в верхний и наоборот. Посчитать сумму цифр в строке.


```python
text = input("Введите строку текста: ")
print("Введено:", text)

longest = ""
current = ""
for ch in text:
    if ch != " ":
        current += ch
    else:
        if len(current) > len(longest):
            longest = current
        current = ""
if len(current) > len(longest):
    longest = current

swapped = ""
for ch in text:
    code = ord(ch)
    if ord("A") <= code <= ord("Z"):
        swapped += chr(code + (ord("a") - ord("A")))
    elif ord("a") <= code <= ord("z"):
        swapped += chr(code - (ord("a") - ord("A")))
    elif ord("А") <= code <= ord("я"):
        if ord("А") <= code <= ord("Я"):
            swapped += chr(code + (ord("а") - ord("А")))
        else:
            swapped += chr(code - (ord("а") - ord("А")))
    elif ch == "Ё":
        swapped += "ё"
    elif ch == "ё":
        swapped += "Ё"
    else:
        swapped += ch

digit_sum = 0
digit_count = 0
for ch in text:
    if "0" <= ch <= "9":
        digit_sum += ord(ch) - ord("0")
        digit_count += 1

print("Самое длинное слово:", longest)
print("Строка с обратным регистром:", swapped)
print("Найдено цифр:", digit_count)
print("Сумма цифр:", digit_sum)
```

**Задание 3.** На вход программе подаётся натуральное число n. Написать программу, которая создаёт список, состоящий из делителей введённого числа. Найти максимальный и минимальный элемент этого списка.


```python
n = int(input("Введите натуральное число n: "))
print("Введено:", n)

if n <= 0:
    print("Число должно быть натуральным")
else:
    divisors = []
    i = 1
    while i <= n:
        if n % i == 0:
            divisors.append(i)
        i += 1

    minimum = divisors[0]
    maximum = divisors[0]
    for value in divisors:
        if value < minimum:
            minimum = value
        if value > maximum:
            maximum = value

    print("Список делителей числа", n, ":", divisors)
    print("Количество делителей:", len(divisors))
    print("Минимальный элемент списка:", minimum)
    print("Максимальный элемент списка:", maximum)
```

**Задание 4.** Найдите три ключа с самыми высокими значениями в словаре `my_dict` = {'a': 12, 'b': 13, 'c': 56, 'd': 400, 'e': 58, 'f': 20}.


```python
my_dict = {'a': 12, 'b': 13, 'c': 56, 'd': 400, 'e': 58, 'f': 20}
print("Исходный словарь:", my_dict)

keys = []
values = []
for key in my_dict:
    keys.append(key)
    values.append(my_dict[key])

count = len(values)
for i in range(count):
    for j in range(count - 1):
        if values[j] < values[j + 1]:
            values[j], values[j + 1] = values[j + 1], values[j]
            keys[j], keys[j + 1] = keys[j + 1], keys[j]

print("Словарь после сортировки по значению (по убыванию):")
for i in range(count):
    print("  ", keys[i], ":", values[i])

print("Три ключа с самыми высокими значениями:")
for i in range(3):
    print("  ", i + 1, "место — ключ", keys[i], "со значением", values[i])
```

**Задание 5.** Реализуйте программу «Магазин игрушек», которая будет включать в себя шесть пунктов меню. У вас есть словарь, где ключ — название игрушки, значение — список, который содержит состав игрушки, цену и количество, которое есть в магазине. Пункты: просмотр описания (название — описание), просмотр цены, просмотр количества, вся информация, покупка (с клавиатуры вводится название игрушки и её количество, n — выход; посчитать цену выбранных товаров и сколько товаров осталось в изначальном списке), до свидания.


```python
toys = {
    "кукла": ["пластик, текстиль, краска", 45.50, 12],
    "машинка": ["металл, пластик, резина", 32.00, 20],
    "конструктор": ["пластик, металл", 78.90, 7],
    "плюшевый мишка": ["текстиль, наполнитель, нитки", 56.30, 9],
}

while True:
    print("\n=== Магазин игрушек ===")
    print("1. Просмотр описания")
    print("2. Просмотр цены")
    print("3. Просмотр количества")
    print("4. Всю информацию")
    print("5. Покупка")
    print("6. До свидания")
    choice = input("Выберите пункт меню: ").strip()
    print("Введено:", choice)

    match choice:
        case "1":
            for name, info in toys.items():
                print(f"{name} – состав: {info[0]}")
        case "2":
            for name, info in toys.items():
                print(f"{name} – цена: {info[1]} руб. за штуку")
        case "3":
            for name, info in toys.items():
                print(f"{name} – в наличии: {info[2]} шт.")
        case "4":
            for name, info in toys.items():
                print(f"{name}: состав — {info[0]}; цена — {info[1]} руб.; количество — {info[2]} шт.")
        case "5":
            total = 0.0
            print("Покупка. Введите название игрушки и количество. n — выход из покупки.")
            while True:
                name = input("Название игрушки (n — выход): ").strip().lower()
                print("Введено:", name)
                if name == "n":
                    break
                if name not in toys:
                    print("Такой игрушки нет в магазине.")
                    continue
                pieces = int(input("Количество (шт.): "))
                print("Введено:", pieces)
                if pieces <= 0:
                    print("Количество должно быть положительным.")
                    continue
                if pieces > toys[name][2]:
                    print(f"Недостаточно товара. Остаток: {toys[name][2]} шт.")
                    continue
                cost = toys[name][1] * pieces
                toys[name][2] -= pieces
                total += cost
                print(f"Добавлено. Стоимость покупки: {cost:.2f} руб. Остаток: {toys[name][2]} шт.")
            print(f"Итоговая стоимость выбранных игрушек: {total:.2f} руб.")
            print("Остаток товаров в магазине:")
            for name, info in toys.items():
                print(f"{name} – {info[2]} шт.")
        case "6":
            print("До свидания")
            break
        case _:
            print("Нет такого пункта меню.")
```

**Задание 6.** Дан кортеж целых чисел. Удалить первый отрицательный элемент.


```python
raw = input("Введите целые числа через пробел: ")
print("Введено:", raw)

numbers = []
current = ""
for ch in raw + " ":
    if ch != " ":
        current += ch
    else:
        if current != "":
            numbers.append(int(current))
            current = ""

source = tuple(numbers)
print("Исходный кортеж:", source)

removed = False
result = ()
for value in source:
    if value < 0 and not removed:
        removed = True
        continue
    result = result + (value,)

if removed:
    print("Первый отрицательный элемент удалён")
else:
    print("Отрицательных элементов нет, кортеж не изменился")
print("Кортеж после обработки:", result)
```
